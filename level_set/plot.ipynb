```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# ---------- Style ----------
rcParams["font.family"] = "serif"
rcParams["mathtext.fontset"] = "dejavuserif"

# ---------- Himmelblau function ----------
def himmelblau(x, y):
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2 - 40

# ---------- Grid ----------
x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)
X, Y = np.meshgrid(x, y)
Z = himmelblau(X, Y)

# ---------- Figure ----------
fig = plt.figure(figsize=(9, 3.6))

# Left: 3D surface
ax1 = fig.add_subplot(1, 2, 1, projection="3d")
surf = ax1.plot_surface(
    X, Y, Z,
    cmap="winter",          # close to the blue-green look
    edgecolor="k",
    linewidth=0.15,
    antialiased=True
)
ax1.view_init(elev=28, azim=-60)
ax1.set_xlim(-5, 5)
ax1.set_ylim(-5, 5)
ax1.set_zlim(-250, 1050)
ax1.set_xticks([-5, 0, 5])
ax1.set_yticks([-5, 0, 5])
ax1.set_zticks([-200, 0, 200, 400, 600, 800, 1000])
ax1.set_xlabel("")
ax1.set_ylabel("")
ax1.set_zlabel("")
ax1.tick_params(labelsize=8)

# Right: 0-level set
ax2 = fig.add_subplot(1, 2, 2)
ax2.contour(X, Y, Z, levels=[0], colors="0.6", linewidths=1.0)
ax2.set_xlim(-5, 5)
ax2.set_ylim(-5, 5)
ax2.set_aspect("equal", adjustable="box")
ax2.set_xticks([-4, -2, 0, 2, 4])
ax2.set_yticks([-4, -2, 0, 2, 4])
ax2.tick_params(labelsize=8)
for spine in ax2.spines.values():
    spine.set_color("0.9")

# Subfigure labels
#fig.text(0.27, 0.06, "(a) Robustness landscape.", ha="center", fontsize=11)
#fig.text(0.73, 0.06, "(b) 0-level set.", ha="center", fontsize=11)

# # Caption
# fig.text(
#     0.5, 0.00,
#     r"Fig. 2. Himmelblau function: $f(x, y) = (x^2 + y - 11)^2 + (x + y^2 - 7)^2 - 40$, with $(-5 \leq x, y \leq 5)$.",
#     ha="center", fontsize=11
# )

plt.subplots_adjust(wspace=0.25, bottom=0.18)
plt.tight_layout(rect=[0, 0.12, 1, 1])

plt.savefig("himmelblau_figure.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

# Style
rcParams["font.family"] = "serif"

# Function
def himmelblau(x, y):
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2 - 40

# Grid
x = np.linspace(-5, 5, 400)
y = np.linspace(-5, 5, 400)
X, Y = np.meshgrid(x, y)
Z = himmelblau(X, Y)

# Figure
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')

# Surface（保留网格线但更细）
surf = ax.plot_surface(
    X, Y, Z,
    cmap="viridis",
    edgecolor='k',       # ✔ 保留网格
    linewidth=0.2,       # ✔ 变细（关键，不然太乱）
    antialiased=True
)

# 🔥 关键：更偏向 z 轴（俯视一点）
ax.view_init(elev=45, azim=-60)
# 你可以试：
# elev=50 更俯视
# elev=40 稍微平一点

# Limits
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_zlim(-250, 1000)

# 轻量化坐标轴（论文风格）
ax.set_xticks([-5, 0, 5])
ax.set_yticks([-5, 0, 5])
ax.set_zticks([-200, 0, 200, 400, 600, 800, 1000])
ax.tick_params(labelsize=8)

# 去掉背景填充（更干净）
ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False

# 可选：让边框更淡
ax.xaxis.pane.set_edgecolor('0.8')
ax.yaxis.pane.set_edgecolor('0.8')
ax.zaxis.pane.set_edgecolor('0.8')

# 保存
plt.tight_layout()
plt.savefig("himmelblau_surface_clear.png", dpi=400, bbox_inches="tight")
plt.show()
```
