```python
import io 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
def _read_csv_text(csv_text: str) -> pd.DataFrame:
    """
    Read CSV from a pasted multiline string.
    """
    return pd.read_csv(io.StringIO(csv_text.strip()))
```

```python
def plot_budget_comparison_3(
    csv_text_1,
    csv_text_2,
    csv_text_3,
    label_1="PARTX",
    label_2="UPART",
    label_3="PBNB",
    title="Cumulative Budget Comparison",
    save_path=None
):
    df1 = _read_csv_text(csv_text_1)
    df2 = _read_csv_text(csv_text_2)
    df3 = _read_csv_text(csv_text_3)

    plt.figure(figsize=(7.5, 5.0))

    plt.plot(df1["iter"], df1["cumulative_budget"],
             marker="o", linewidth=2.5, label=label_1)

    plt.plot(df2["iter"], df2["cumulative_budget"],
             marker="s", linewidth=2.5, label=label_2)

    plt.plot(df3["iter"], df3["cumulative_budget"],
             marker="^", linewidth=2.5, label=label_3)

    plt.xlabel("Iteration")
    plt.ylabel("Cumulative Budget")
    plt.title(title)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300)
    plt.show()
```

```python
def plot_maintain_line_3(
    csv_text_1,
    csv_text_2,
    csv_text_3,
    label_1="PARTX",
    label_2="UPART",
    label_3="PBNB",
    delta=0.1,
    title="Maintain Region Ratio Comparison",
    save_path=None
):
    df1 = _read_csv_text(csv_text_1)
    df2 = _read_csv_text(csv_text_2)
    df3 = _read_csv_text(csv_text_3)

    plt.figure(figsize=(7.5, 5.0))

    plt.plot(df1["iter"], df1["ratio_M"],
             marker="o", linewidth=2.5, label=label_1)

    plt.plot(df2["iter"], df2["ratio_M"],
             marker="s", linewidth=2.5, label=label_2)

    plt.plot(df3["iter"], df3["ratio_M"],
             marker="^", linewidth=2.5, label=label_3)

    # 🔥 δ = 0.1 reference line
    plt.axhline(
        y=delta,
        color='black',
        linestyle='--',
        linewidth=2,
        label=r'$\delta = 0.1$'
    )

    plt.xlabel("Iteration")
    plt.ylabel("Maintain Region Ratio ($\\Theta^-$)")
    plt.title(title)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300)
    plt.show()
```
